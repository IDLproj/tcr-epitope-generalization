```python
import pandas as pd

#Print summary statistics for each split
for split in ["train", "val", "test"]:
    df = pd.read_csv(f"../splits/{split}.csv")

    print(f"\n{split.upper()}")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist())
    print("Missing values:", df.isnull().sum().to_dict())
    print("B3 lengths:", df["B3"].str.len().min(), df["B3"].str.len().max())
    print("Peptide lengths:", df["peptide"].str.len().min(), df["peptide"].str.len().max())
    print("Binding distribution:", df["binder"].value_counts().to_dict())
```
